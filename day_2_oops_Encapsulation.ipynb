```python
class Point:

  def __init__(self,x,y):
    self.x_cod = x
    self.y_cod = y
  def __str__(self):
    return '<{},{}>'.format(self.x_cod,self.y_cod)

  def eucliden_distance(self,other):
    return ((self.x_cod - other.x_cod)**2 + (self.y_cod - other.y_cod)**2)**0.5

  def distance_from_origin(self):
    return self.eucliden_distance(Point(0,0))


class Line:

  def __init__(self,A,B,C):
    self.A = A
    self.B = B
    self.C = C

  def __str__(self):
    b_sign = '+' if self.B >= 0 else '-'
    c_sign = '+' if self.C >= 0 else '-'
    return '{}x {} {}y {} {} = 0'.format(
        self.A, b_sign, abs(self.B), c_sign, abs(self.C)
      )

  def point_on_line(line,point):
    if line.A*point.x_cod + line.B*point.y_cod + line.C == 0:
      return 'point lies on the line'

    else:
      return 'point does not lie on the line'

  def shortest_distance(line,point):
    return abs(line.A*point.x_cod + line.B*point.y_cod + line.C)/(line.A**2 + line.B**2)**0.5


```

```python
l1 = Line(1,1,-2)
p1 = Point(1,1)
print(l1)
print(p1)
print(l1.point_on_line(p1))
print(l1.shortest_distance(p1))
```

# **REFERENCE VARIABLE**

*    Reference variable hold the object memory address
*    We can create object without variable as well
*    An object can have multiple
*    Assigning a new reference variable to an existing objects does not create a new object




```python
class Person:
  def __init__(self):
    self.name = "shikhar"
    self.age = 23

p = Person()
q = p
```

```python
# Multiple reference
print(id(p))
print(id(q))

# both are same because point to the same memory address
```

```python
#  Change attribute value with using of second object
```

```python
print(p.name)
print(q.name)
q.name = "sharma"
print(p.name)
print(q.name)
```

**We can pass the class object to an function as argument**

```python
class Person:
  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

# function is outside of the class
def greet(self):
  print ('Hi my name is' ,self.name,'and i am', self.gender)

  # also function can give you an object
  p1 = Person("modiji","male")
  return p1

p = Person("shikhar","male")
# now calling the function and passing the object reference which is making of Person class
x = greet(p)
# now print the object which are given by the function called greet
print(x.name)
print(x.gender)
```

```python
#User define objects are mutable in python because we change the value in same memory location
```

```python
class Person:
  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

# function is outside of the class
def greet(self):
  self.name = "nakamura"
  return self

p = Person("shikhar","male")
print(id(p))
p1 = greet(p)
print(id(p1))
```

# *Collection of objects*

```python
class Person:
  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

p1 = Person("shikhar","male")
p2 = Person("meloni","female")
p3 = Person("modiji","male")

l = [p1,p2,p3]
# print each object memory address
print(l)
for i in l:
  print(i.name , i.gender)
```

```python
# working with discionary as well

class Person:
  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

p1 = Person("shikhar","male")
p2 = Person("meloni","female")
p3 = Person("modiji","male")

d = {'p1':p1,'p2':p2,'p3':p3}
# print each object memory address
print(d)
for i in d:
  print(d[i].name,d[i].gender)
```

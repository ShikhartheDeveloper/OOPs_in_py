```python
class Point:

  def __init__(self,x,y):
    self.x_cod = x
    self.y_cod = y
  def __str__(self):
    return '<{},{}>'.format(self.x_cod,self.y_cod)

  def eucliden_distance(self,other):
    return ((self.x_cod - other.x_cod)**2 + (self.y_cod - other.y_cod)**2)**0.5

  def distance_from_origin(self):
    return self.eucliden_distance(Point(0,0))


class Line:

  def __init__(self,A,B,C):
    self.A = A
    self.B = B
    self.C = C

  def __str__(self):
    b_sign = '+' if self.B >= 0 else '-'
    c_sign = '+' if self.C >= 0 else '-'
    return '{}x {} {}y {} {} = 0'.format(
        self.A, b_sign, abs(self.B), c_sign, abs(self.C)
      )

  def point_on_line(line,point):
    if line.A*point.x_cod + line.B*point.y_cod + line.C == 0:
      return 'point lies on the line'

    else:
      return 'point does not lie on the line'

  def shortest_distance(line,point):
    return abs(line.A*point.x_cod + line.B*point.y_cod + line.C)/(line.A**2 + line.B**2)**0.5


```

```python
l1 = Line(1,1,-2)
p1 = Point(1,1)
print(l1)
print(p1)
print(l1.point_on_line(p1))
print(l1.shortest_distance(p1))
```

# **REFERENCE VARIABLE**

*    Reference variable hold the object memory address
*    We can create object without variable as well
*    An object can have multiple
*    Assigning a new reference variable to an existing objects does not create a new object




```python
class Person:
  def __init__(self):
    self.name = "shikhar"
    self.age = 23

p = Person()
q = p
```

```python
# Multiple reference
print(id(p))
print(id(q))

# both are same because point to the same memory address
```
